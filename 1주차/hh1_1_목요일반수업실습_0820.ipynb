```python
# crx.data를 읽으셔서, 컬럼명은 A1~A16이 되도록 지정해주세요
df12 = pd.read_csv("crx.data", header=None) 
df12.columns = ['A'+str(i) for i in range(16) ]
df12["A15"] = df12.A15.replace( {'+':0, '-':1})# 16번째 컬럼이 +,- -> replace를 이용하셔서 1,0으로 변경해주세요.
df12.drop(columns="A13", inplace=True)# 14번째 컬럼 : 삭제
df12["A15"] = df12.A15.astype("int")# type점검->적절한 type
df12_1 = df12.select_dtypes(include="number")
df12_2 = df12.select_dtypes(exclude="number")

# 수치형-> imp, scaling(standard)
imp_n = SimpleImputer()
scaler_n = StandardScaler()
scaled = scaler_n.fit_transform( imp_n.fit_transform(df12_1) )
df12_1_1 = pd.DataFrame(scaled, columns=df12_1.columns)
#범주형->imp, get_dummies로 OHE
imp_c = SimpleImputer(strategy="most_frequent")  
imputed = imp_c.fit_transform(df12_2)
df12_2_imp = pd.DataFrame( imputed, columns=df12_2.columns)
df12_2_2 = pd.get_dummies(df12_2_imp)

df12_3 = pd.concat(  [df12_1_1, df12_2_2   ], axis=1 )# 하나의 데이터프레임으로 합쳐주세요. (pd.concat)
train, test = train_test_split(df12_3, train_size=0.9, stratify=df12_3['A15'])
train.to_csv("traintrain.csv", index=False)
```
