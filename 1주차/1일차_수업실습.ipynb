```python
import pandas as pd
df2 = pd.read_csv("binary.csv")
df2.head() 
df2.values
#df2.info() #데이터 소개
df2.describe()  #전체 컬럼 요약
df2["admit"].value_counts() #컬럼 고유한 값 카운트
```

```python
#data.csv를 데이터프레임으로 만드세요.passorfail컬럼에대해 value_counts 적용
df3 = pd.read_csv("data.csv")#sep, header, skiprows, index_col..
#df3["Pass.Fail"].value_counts()
df3[  ["X20", "X86"]  ]  
df3.iloc[  :  , 0:2   ]   #integer location indexing / slicing 시작위치:끝나는위치
```

```python
# 1: query,  Admit 컬럼이 1인 행들만 출력해보세요.
df2_1 = df2.query("admit == 1 & rank >=3") #and
#df2.query("admit == 1 | rank >=3") #or 
#df999.query("admit == '1' ") # 문자열 조건비교
df2_1.to_csv("result.csv", index=False) 
```

```python
# 2:  sample, 
df2.sample(10) 

#df2에서 admit== 0 사람 중에 rank가 1이나 2인 사람들의 gre, gpa 평균
df2.query("admit == 0 & rank <=2")[["gre","gpa"]].mean() 
```

```python
# 3: groupby :  pivot, pivot_table, aggregate...
df2.groupby(["admit", "rank"]).mean()  #mean, max, min, median

#df3에 Pass.Fail을 기준으로 X20, X86 평균을 비교하세요.
df3.groupby("Pass.Fail")[["X20", "X86"]].mean()

df3["C1"] = df3.X20 > df3.X20.mean() #컬럼 생성
df3["C1"] = df3.C1.astype("int")  #type변경

df3.groupby(["Pass.Fail", "C1"])[["X86"]].mean()
df3.drop(columns="C1") #컬럼 지우기
```

```python
#4. replace 
# train_df.csv 를 df4로 읽어보세요, (encoding="cp949"를 적용)
# info로 파악해보세요.
df4 = pd.read_csv("train_df.csv", encoding="cp949")
df4.head(1)
df4["발화자2"] = df4.발화자.replace( {"c":"Customer", "s":"Support"  } )   
df4["발화자2"].str.replace({"S":"s", "C":"c"}) #부분적으로 일치되는 것도 치환
df4["발화자2"].str.replace("[t-x]", "0", regex=True   ) #정규표현식
#정규표현식: [a-z], [A-Z], [a-e], [가-힣]
#df4.인텐트  에서 [가-잉] 에 해당하는 문자를 a로 치환해보세요.
df4["인텐트"].str.replace("[가-잉]", "a", regex=True   )
```

```python
#5. apply, applymap, np.where( df4.상담내순번>2,"late", "early")
def func1(x):
    x_len = len(x)
    if x_len <= 10:
        return 1
    elif x_len<=20:
        return 2
    else:
        return 3  
df4['발화문'].apply(func1) 
# 발화문 컬럼의 길이를 측정해서 ~10 1, ~20 2, 21~ 3
# 길이측정은 len(문자열)
```

```python
# 6. merge, concat
# 두 테이블 join -> 기준 컬럼
#pd.merge( df1, df2, how="left", on="기준컬럼")
#pd.concat( [df1, df2, df3], axis=1또는0 )
df2.shape
pd.concat([df2, df2], axis=1).shape #열1
df4_1 = df4.iloc[0:10, 0:2]
df4_2 = df4.iloc[2:12, 0:3]
pd.merge( df4_1, df4_2, how="left", on="IDX")
```

```python
#7. 결측치 확인
df5 = pd.read_csv("creditApprovalUCI.csv")
df5.isnull().mean().sort_values(ascending=False) #결측율
df5.dropna() #결측이 있는 행 삭제, inplace=True는 원본에 반영
df5.fillna(0) #dtype 기준으로 나눠서 작업
df5.select_dtypes(exclude="str")  #exclude

#df5에서 수치형 변수들만 선택->0으로 imputing
df5_1 = df5.select_dtypes(exclude="str").fillna(0)  

#df5에서 범주형 변수들만 선택->"unknown"으로 imputing
df5_2 = df5.select_dtypes(include="str").fillna("unknown")

#하나로 합치셔서 df6을 만들어보세요 (pd.concat, pd.merge)
pd.concat([df5_1, df5_2], axis=1)
```

```python
from sklearn.impute import SimpleImputer
df5 = pd.read_csv("creditApprovalUCI.csv")
df5_1 = df5.select_dtypes(exclude="str")  
df5_2 = df5.select_dtypes(include="str")

imputer = SimpleImputer(strategy="mean") #median
df5_1 = pd.DataFrame( imputer.fit_transform(df5_1), columns=df5_1.columns)  

imputer2 = SimpleImputer(strategy="most_frequent")
df5_2 = pd.DataFrame( imputer2.fit_transform(df5_2), columns=df5_2.columns)  

pd.concat( [ df5_1, df5_2], axis=1)
```
