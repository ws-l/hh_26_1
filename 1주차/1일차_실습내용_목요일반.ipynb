```python
# 7 결측치 체크, 채우기, imputing : NaN (not a number), np.nan
# missing value: NA
# creditApprovalUCI.csv를 읽어서 df13을 만드세요
df13 = pd.read_csv("creditApprovalUCI.csv")
df13.isnull().mean().sort_values(ascending=False)
df13.dropna() #inplace=True 원본에 반영시킬때, #결측이 하나라도 있는 행은 삭제
#수치형, 수치 아닌형으로 컬럼을 분리
df13.select_dtypes(include="number").fillna(0)
df13.select_dtypes(exclude="number").fillna("unk")
```

```python
#!pip install scikit-learn -q
from sklearn.impute import SimpleImputer
df13_num = df13.select_dtypes(include="number")
df13_cat = df13.select_dtypes(exclude="number")

imputer = SimpleImputer() #수치mean, median, 수치아닌것 strategy="most_frequent"
imputed = imputer.fit_transform(df13_num) #imputer.fit(데이터프레임) #imputer.transform(데이터프레임)
df13_num_imp = pd.DataFrame(imputed, columns=df13_num.columns)

#범주형(수치아닌형) 변수들에 대한 imputing하셔서 결과를 데이터프레임으로 만들어보세요.
imputer2 = SimpleImputer( strategy="most_frequent" )
imputed2 = imputer2.fit_transform(df13_cat)
df13_cat_imp = pd.DataFrame(imputed2, columns=df13_cat.columns)

#수치형 결과, 범주형결과를 하나의 데이터프레임으로 합쳐보세요.  
pd.concat([df13_num_imp, df13_cat_imp] , axis=1)   
```

```python
# 8 scaling
from sklearn.preprocessing import MinMaxScaler, StandardScaler
scaler = StandardScaler()
scaled = scaler.fit_transform( df13_num_imp )  
pd.DataFrame( scaled, columns = df13_num_imp.columns  )

#creditset2.csv -> age, loan, income 선택하셔서 MinMaxScaling해보세요.
df14 = pd.read_csv("creditset2.csv")
df14 = df14[["age", "loan", "income"]]
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df14)
pd.DataFrame( scaled, columns=df14.columns)   
```
